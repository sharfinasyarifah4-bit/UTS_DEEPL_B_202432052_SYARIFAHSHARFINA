  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;"> Pada bagian ini digunakan untuk memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432052_SyarifahSharfina

%pip install torch torchvision torchaudio 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# Seed agar hasil pelatihan dapat direproduksi
np.random.seed(42)
torch.manual_seed(42)
print("PyTorch versi:", torch.__version__)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;"> Pada bagian ini, saya memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432052_SyarifahSharfina
# 1. Fungsi untuk memuat satu file CSV lalu merapikannya
def muat_data(path):
    df = pd.read_csv(path, parse_dates=['date'])
    return df[['date', 'meantemp']].sort_values('date').reset_index(drop=True)

# 2. Muat data latih dan data uji
df_train = muat_data('DailyDelhiClimateTrain.csv')
df_test = muat_data('DailyDelhiClimateTest.csv')

# 3. Eksplorasi singkat dataset
print("Ukuran data latih :", df_train.shape)
print("Ukuran data uji   :", df_test.shape)
print("Rentang data latih:", df_train['date'].min().date(), "s/d", df_train['date'].max().date())
print("Rentang data uji  :", df_test['date'].min().date(), "s/d", df_test['date'].max().date())
print("Nilai kosong      :", int(df_train['meantemp'].isna().sum() + df_test['meantemp'].isna().sum()))
print(df_train.head())
print(df_train['meantemp'].describe().round(2))

# 4. Visualisasi deret waktu suhu harian
plt.figure(figsize=(12, 4))
plt.plot(df_train['date'], df_train['meantemp'], color='#114F6A', linewidth=1, label='Data Latih')
plt.plot(df_test['date'], df_test['meantemp'], color='#1FA4B1', linewidth=1.5, label='Data Uji')
plt.title('Suhu Rata-rata Harian Delhi (2013 - 2017)')
plt.xlabel('Tanggal')
plt.ylabel('Suhu (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
#202432052_SyarifahSharfina
# 1. Parameter sliding window: 14 hari terakhir dipakai untuk memprediksi hari berikutnya
WINDOW = 14

# 2. Ambil kolom target (suhu) sebagai array 2 dimensi
train_vals = df_train['meantemp'].values.reshape(-1, 1)
test_vals = df_test['meantemp'].values.reshape(-1, 1)

# 3. Normalisasi Min-Max: parameter HANYA dihitung dari data latih (mencegah data leakage)
t_min, t_max = train_vals.min(), train_vals.max()
def normalisasi(x):
    return (x - t_min) / (t_max - t_min)
def denormalisasi(x):
    return x * (t_max - t_min) + t_min

train_scaled = normalisasi(train_vals)
test_scaled = normalisasi(test_vals)

# 4. Fungsi pembentuk sekuens (sliding window)
def create_sequences(data, window):
    X, y = [], []
    for i in range(len(data) - window):
        X.append(data[i:i + window])
        y.append(data[i + window])
    return np.array(X), np.array(y)

# 5. Data uji diberi konteks WINDOW hari terakhir dari data latih agar prediksi hari pertama valid
test_input = np.vstack([train_scaled[-WINDOW:], test_scaled])

X_train, y_train = create_sequences(train_scaled, WINDOW)
X_test, y_test = create_sequences(test_input, WINDOW)

# 6. Konversi ke PyTorch Tensor bertipe float32
X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.float32)
X_test = torch.tensor(X_test, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.float32)

# 7. Bungkus data latih ke DataLoader (mini-batch 32)
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=32, shuffle=True)

print("X_train:", tuple(X_train.shape), "| y_train:", tuple(y_train.shape))
print("X_test :", tuple(X_test.shape), "| y_test :", tuple(y_test.shape))
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Pada bagian ini, saya mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432052_SyarifahSharfina
class SequenceModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, num_layers=2, dropout=0.2):
        super(SequenceModel, self).__init__()
        # Lapisan LSTM bertumpuk 2 layer; batch_first=True -> input (batch, waktu, fitur)
        self.lstm = nn.LSTM(input_size=input_size, hidden_size=hidden_size,
                            num_layers=num_layers, batch_first=True, dropout=dropout)
        # Lapisan Linear untuk memetakan hidden state terakhir menjadi 1 nilai suhu
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, (h_n, c_n) = self.lstm(x)       # out: hidden state di semua time step
        return self.fc(out[:, -1, :])        # ambil time step terakhir lalu prediksi

print(SequenceModel())
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Pada bagian ini, saya menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432052_SyarifahSharfina
# 1. Inisiasi hyperparameter
epochs = 50
learning_rate = 0.01

# 2. Inisiasi model, fungsi loss, dan optimizer
model = SequenceModel(input_size=1, hidden_size=64, num_layers=2, dropout=0.2)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# 3. Riwayat loss untuk divisualisasikan
train_losses = []
test_losses = []

# 4. Training loop
for epoch in range(epochs):
    model.train()
    total_loss = 0.0
    for xb, yb in train_loader:
        optimizer.zero_grad()                 # reset gradien
        pred = model(xb)                      # forward propagation
        loss = criterion(pred, yb)            # hitung MSE
        loss.backward()                       # backpropagation (BPTT)
        optimizer.step()                      # perbarui bobot (Adam)
        total_loss += loss.item() * xb.size(0)
    train_losses.append(total_loss / len(X_train))

    model.eval()
    with torch.no_grad():
        test_losses.append(criterion(model(X_test), y_test).item())

    if (epoch + 1) % 5 == 0 or epoch == 0:
        print(f"Epoch [{epoch+1:2d}/{epochs}] | Train Loss: {train_losses[-1]:.5f} | Test Loss: {test_losses[-1]:.5f}")
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Pada bagian ini, digunakan untuk menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432052_SyarifahSharfina
# 1. Plot Grafik Penurunan Loss Training terhadap Epoch
plt.figure(figsize=(10, 4))
plt.plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.plot(test_losses, color='#1FA4B1', linewidth=2, linestyle='--', label='Test Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE, skala ternormalisasi)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 2. Prediksi pada data uji lalu kembalikan ke skala suhu asli (°C)
model.eval()
with torch.no_grad():
    pred_scaled = model(X_test).numpy()
y_pred = denormalisasi(pred_scaled).flatten()
y_true = denormalisasi(y_test.numpy()).flatten()

# 3. Metrik evaluasi
mae = np.mean(np.abs(y_true - y_pred))
rmse = np.sqrt(np.mean((y_true - y_pred) ** 2))
r2 = 1 - np.sum((y_true - y_pred) ** 2) / np.sum((y_true - y_true.mean()) ** 2)
print(f"MAE  : {mae:.3f} °C")
print(f"RMSE : {rmse:.3f} °C")
print(f"R²   : {r2:.4f}")

# 4. Plot Grafik Perbandingan Data Aktual vs Hasil Prediksi Model
plt.figure(figsize=(12, 5))
plt.plot(df_test['date'], y_true, color='#114F6A', linewidth=2, label='Data Aktual')
plt.plot(df_test['date'], y_pred, color='#1FA4B1', linewidth=2, linestyle='--', label='Prediksi LSTM')
plt.title('Perbandingan Suhu Aktual vs Prediksi LSTM pada Data Uji')
plt.xlabel('Tanggal')
plt.ylabel('Suhu (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 5. Sebaran prediksi vs aktual (titik ideal berada pada garis diagonal)
plt.figure(figsize=(5, 5))
plt.scatter(y_true, y_pred, color='#1FA4B1', alpha=0.7)
batas = [min(y_true.min(), y_pred.min()), max(y_true.max(), y_pred.max())]
plt.plot(batas, batas, color='#114F6A', linestyle='--', label='Prediksi ideal')
plt.title('Sebaran Aktual vs Prediksi')
plt.xlabel('Aktual (°C)')
plt.ylabel('Prediksi (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
